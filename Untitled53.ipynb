```python

import numpy as np


hours = np.array([1, 2, 3, 4, 5], dtype=float)
marks = np.array([20, 30, 40, 50, 60], dtype=float)


m = 0
b = 0

learning_rate = 0.01
epochs = 1000
n = len(hours)

# Cost Function
def cost_function(y_actual, y_predicted):
    return np.mean((y_actual - y_predicted) ** 2)

# Gradient Descent
for i in range(epochs):

    # Predict marks
    predicted = m * hours + b

    # Calculate gradients
    dm = (2/n) * np.sum(hours * (predicted - marks))
    db = (2/n) * np.sum(predicted - marks)

    # Update parameters
    m = m - learning_rate * dm
    b = b - learning_rate * db

# Final predictions
final_predictions = m * hours + b

# Calculate final cost
final_cost = cost_function(marks, final_predictions)

# Predict for 6 study hours
new_hours = 6
predicted_marks = m * new_hours + b

# Display results
print("Slope:", round(m, 2))
print("Intercept:", round(b, 2))
print("Final Cost:", round(final_cost, 4))
print("Predicted Marks:", round(predicted_marks, 2))
```
